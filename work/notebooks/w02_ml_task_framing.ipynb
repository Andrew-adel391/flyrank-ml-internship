## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why ML: Fixed if/else rules (e.g., checking for specific keywords or character counts) fail because natural language domain inputs are highly variable. A fixed rule breaks whenever terminology varies or when complex context is split across short phrases. ML captures subtle semantic context and feature interactions that static rules completely miss, without requiring constant manual rule updates

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Demonstrate fixed rule failure vs pattern variability
sample_inputs = [
    "Need clinical guidance on dosage",
    "Where is the guide for dosage?"
]

# Fixed rule check (naive keyword matching)
rule_match = [1 if "guidance" in text else 0 for text in sample_inputs]
print(f"Naive Rule Matches: {rule_match} (Misses semantic equivalence)")
```
