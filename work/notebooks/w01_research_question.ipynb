## 2. The question: decision, action, cost of a wrong call

* **Research Question:** Which existing content pages should be prioritized for immediate intent re-alignment or content refresh to capture maximum potential click growth over the next quarter?

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

### Data Highlights from Starter Dataset:

* **Observed Dataset Volume:** Analyzed **78,835,655** daily performance records from the `fact_content_daily_performance` dataset via DuckDB streaming.
* **Zero-Click Opportunity:** Exactly **95.93%** of daily impression records resulted in **0 clicks**, revealing a massive surface area where search intent is uncaptured.
* **Baseline Engagement Friction:** Across all records, the average daily performance shows **22.40 impressions** versus only **0.08 clicks** per record, indicating significant position and CTR friction across un-optimized pages.

```python
import duckdb

# 1. Connect and initialize HuggingFace credentials
con = duckdb.connect()
# Note: Ensure you accept the repository gate on HuggingFace first, then insert your token below
con.execute("CREATE SECRET (TYPE huggingface, TOKEN '*************')")

rel = "hf://datasets/FlyRank/internship-warehouse"

# 2. Execute metric queries
dataset_path = f"{rel}/fact_content_daily_performance/**/*.parquet"

total_rows = con.sql(f"SELECT COUNT(*) FROM read_parquet('{dataset_path}')").fetchone()[0]

stats = con.sql(f"""
    SELECT
        COUNT(CASE WHEN gsc_clicks = 0 THEN 1 END) * 100.0 / COUNT(*) as zero_click_pct,
        AVG(gsc_impressions) as avg_impressions,
        AVG(gsc_clicks) as avg_clicks
    FROM read_parquet('{dataset_path}')
""").fetchone()

print(f"Total Observed Performance Records: {total_rows:,}")
print(f"Zero-Click Opportunity Rate: {stats[0]:.2f}%")
print(f"Average Daily Impressions per Record: {stats[1]:.2f}")
print(f"Average Daily Clicks per Record: {stats[2]:.2f}")
```
